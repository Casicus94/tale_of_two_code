# T3. Best-equation-first walk for Medellín

T3 (B7): best-equation-first walk for Medellin. Writes variable x complexity 0/1
matrices and per-variable persistence summaries for each Medellin front, and prints the
robust core (variables continuous from first appearance to the top of the front).

Fronts used: stored explanatory fronts (PM2.5: 2026-09-10 15:11, after the v11 master;
O3: 2026-09-10 15:12, which predates the final O3 rebuild -- so the O3 walk is repeated
on medellin_o3_pysr_pareto_front_refresh.csv, re-run on the current selected_no_lag
features), plus the lagged-only fronts.

```python
# Setup
import sys
from pathlib import Path
HERE = Path.cwd().resolve()          # notebooks/ folder
sys.path.insert(0, str(HERE.parent)); sys.path.insert(0, str(HERE.parents[1]))
from pathlib import Path
import sys
import pandas as pd
from termmap_utils import DATA, load_front, term_presence, term_summary
from style import *   # PM25, O3, NAVY, GREY, RED, LIGHT, ROOT, DATA_OUT, save_fig, panel_label
```

```python
# Analysis
OUT = ROOT / "Paper" / "figures" / "data"
M = DATA / "medellin"
FRONTS = {
    "medellin_pm25": M / "medellin_pysr_pareto_front.csv",
    "medellin_o3": M / "medellin_o3_pysr_pareto_front.csv",
    "medellin_o3_refresh": M / "medellin_o3_pysr_pareto_front_refresh.csv",
    "medellin_pm25_forecast": M / "medellin_pm25_forecast_interpretable_pysr_pareto.csv",
    "medellin_o3_forecast": M / "medellin_o3_forecast_interpretable_pysr_pareto.csv",
}
for name, path in FRONTS.items():
    f = load_front(path)
    mat = term_presence(f)
    mat.to_csv(OUT / f"{name}_term_presence.csv")
    summ = term_summary(mat)
    summ.to_csv(OUT / f"{name}_term_summary.csv", index=False)
    print(f"\n== {name}  (complexities {f.complexity.min()}-{f.complexity.max()}, {len(f)} equations)")
    print(summ.sort_values(["first_complexity"]).to_string(index=False))
```
